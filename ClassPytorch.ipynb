```python
import torch
import torch.nn as nn
import numpy as np
```

```python
np.random.seed(101)
xx = np.linspace(3,8,11)
yy = xx**2 + 1 + np.random.randn(11)
```

```python
xx
```

```python
xx = xx.reshape(-1,1)
```

```python
xx = torch.FloatTensor(xx)
yy = torch.FloatTensor(yy)
```

```python
from torch.optim import SGD, Adam
```

```python
modl = nn.Sequential(nn.Linear(1,1))
```

```python
# modl(xx)
```

```python
lossfn1 = nn.MSELoss()
# sgd = SGD(modl.parameters(), lr=0.01)
sgd = Adam(modl.parameters(), lr=0.1)
```

```python
for _ in range(100):
  sgd.zero_grad()
  y_pred = modl(xx)
  loss = lossfn1(y_pred, yy)
  loss.backward()
  sgd.step()
  print(loss.item())
```

```python
from sklearn.datasets import load_iris
```

```python
dt = load_iris()
```

```python
x = dt.data
y = dt.target
```

```python
x = torch.FloatTensor(x)
y = torch.LongTensor(y)
```

```python
class Model(nn.Module):
    def __init__(sf):
        super().__init__()
        sf.layer1 = nn.Linear(4, 10)
        sf.layer2 = nn.Linear(10, 3)
        # sf.sx = nn.Softmax(dim=1)

    def forward(sf, x):
        x = sf.layer1(x)
        x = torch.relu(x)
        x = sf.layer2(x)
        # return x.shape
        # x = sf.sx(x)
        return x
```

```python
import torch
```

```python
mod = Model()
```

```python
lossfn = nn.CrossEntropyLoss()
op = Adam(mod.parameters(), lr=0.01)
```

```python
for i in range(10):
    op.zero_grad()
    y_pred = mod(x)
    loss = lossfn(y_pred, y)
    loss.backward()
    op.step()
    print(loss.item())
```

```python
import torch
import torch.nn as nn
```

```python
class Birds(nn.Module):
  def __init__(self):
    super().__init__()

    self.y = nn.Linear(10, 2)
    self.x = nn.Linear(5, 10)
    self.r = nn.ReLU()

  def forward(self, d):
    d = self.x(d)
    d = self.r(d)
    d = self.y(d)
    return d.shape
```
